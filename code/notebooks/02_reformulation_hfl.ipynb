## Iteration 2 — сравнение эквивалентных формулировок одной задачи (HFL-consensus)

В статье (arXiv:2602.04479) обсуждается общий класс задач (P) с аффинными связующими ограничениями и то,
что **оценки сложности алгоритмов могут зависеть не только от “самой задачи”, но и от формы её записи**.

Здесь мы делаем минимальный эксперимент на toy-задаче Horizontal Federated Learning (HFL):
каждый агент $$i$$ имеет локальную квадратичную функцию
$$
f_i(x) = \frac12 \|x - b_i\|^2,
$$
и хотим прийти к **общему решению** (консенсус).

Мы сравним две **математически эквивалентные** записи одной и той же задачи:

1) **Shared-variable consensus**: $$x_i = z$$ для всех агентов.  
2) **Edge / coupled constraints** на графе коммуникации $$G=(V,E)$$:
   $$x_i - x_j = 0 \quad \forall (i,j)\in E.$$

Для обеих форм запускаем ADMM и сравниваем:
- число итераций до заданной точности по **ошибке консенсуса**,
- как результат зависит от топологии графа.


```python
import sys, os
from pathlib import Path
import importlib

print("python:", sys.executable)
print("cwd:", os.getcwd())

# Кандидат на корень репо (нужно, чтобы там лежала папка 'code')
ROOT = Path.cwd().resolve().parents[2]
print("ROOT:", ROOT)
print("ROOT exists:", ROOT.exists())
print("ROOT/code exists:", (ROOT / "code").exists())
print("ROOT/code/__init__.py exists:", (ROOT / "code" / "__init__.py").exists())

# Вставим ROOT в sys.path (на всякий случай)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("sys.path[0]:", sys.path[0])

# Посмотрим, что такое 'code' при импорте
m = importlib.import_module("code")
print("imported module 'code' from:", getattr(m, "__file__", "<built-in>"))
```
